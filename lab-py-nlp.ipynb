## **2. Text Preprocessing**

### **Exercise 1: Tokenization and Stop Word Removal**

Tokenize the following text

```python
text = "Natural Language Processing (NLP) is a fascinating field of study! It involves analyzing and understanding human language."

# your code here
tokens = word_tokenize(text)

print("Tokens:", tokens)
print("number of tokens:", len(tokens))
```

Remove stop words and store the result in a variable called `filtered_tokens`

```python
stop_words = set(stopwords.words('english'))

# your code here
# .lower() matters: NLTK's stop-word list is lowercase, so "It" would survive without it
filtered_tokens = [w for w in tokens if w.lower() not in stop_words]

print("removed:", [w for w in tokens if w.lower() in stop_words])
print("%d tokens -> %d after stop-word removal" % (len(tokens), len(filtered_tokens)))
```

```python
print("Filtered Tokens:", filtered_tokens)
```

### **Exercise 2: Stemming and Lemmatization**

Apply stemming and lemmatization to the `filtered_tokens`. Compare the results.

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

Apply stemming and store the result in `stemmed_tokens`

```python
# your code here
# Stemming: chop affixes off with rules. Fast, crude, and the result is often
# not a real word ("study" -> "studi").
stemmed_tokens = [stemmer.stem(w) for w in filtered_tokens]
```

```python
print("Stemmed Tokens:", stemmed_tokens)
```

Apply lemmatization and store the result in `lemmatized_tokens`

```python
# your code here
# Lemmatization: look the word up in WordNet and return its dictionary form.
# Slower, but the output is always a real word.
lemmatized_tokens = [lemmatizer.lemmatize(w) for w in filtered_tokens]

# --- side by side ---
print("%-16s %-16s %s" % ("ORIGINAL", "STEMMED", "LEMMATIZED"))
for o, s, l in zip(filtered_tokens, stemmed_tokens, lemmatized_tokens):
    print("%-16s %-16s %s" % (o, s, l))

# --- why lemmatization looks like it did almost nothing ---
# .lemmatize() assumes the word is a NOUN unless told otherwise, so verbs are
# left alone. Tell it the part of speech and the behaviour changes completely:
print()
print("with pos='v' (treat as verbs):")
for w in ["involves", "analyzing", "understanding", "studies"]:
    print("  %-15s noun-default -> %-15s pos='v' -> %s"
          % (w, lemmatizer.lemmatize(w), lemmatizer.lemmatize(w, pos='v')))
```

```python
print("Lemmatized Tokens:", lemmatized_tokens)
```

## **3. Feature Extraction**

### **Exercise 3: Bag of Words (BoW)**

Use the `CountVectorizer` from `scikit-learn` to create a Bag of Words representation of the following corpus

```python
corpus = [
    "I love NLP.",
    "NLP is amazing.",
    "I enjoy learning new things in NLP."
]
```

```python
# your code here
# Step 1: Initialize the CountVectorizer
vectorizer = CountVectorizer()

# Step 2: Fit and transform the corpus into a BoW representation
X = vectorizer.fit_transform(corpus)

# --- readable view: one row per document, one column per word ---
print("vocabulary:", vectorizer.get_feature_names_out())
print("shape:", X.shape, " (3 documents x 9 words)")
print()
print(pd.DataFrame(X.toarray(),
                   columns=vectorizer.get_feature_names_out(),
                   index=["doc0", "doc1", "doc2"]).to_string())

# --- note: CountVectorizer does its OWN tokenizing ---
# It lowercases, strips punctuation, and by default keeps only tokens of
# 2+ word characters - so the standalone "I" in doc0 and doc2 is dropped.
# None of the NLTK work above is used here; sklearn starts from the raw strings.
```

```python
print("Bag of Words:\n", X.toarray())
print("Vocabulary:", vectorizer.get_feature_names_out())
```

### **Exercise 4: TF-IDF**

Use the `TfidfVectorizer` from `scikit-learn` to create a TF-IDF representation of the same corpus. Store the result in `X_tfidf`

```python
# your code here
# Step 1: Initialize the TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

# Step 2: Fit and transform the corpus into a TF-IDF representation
X_tfidf = tfidf_vectorizer.fit_transform(corpus)

# --- readable view ---
print(pd.DataFrame(X_tfidf.toarray().round(3),
                   columns=tfidf_vectorizer.get_feature_names_out(),
                   index=["doc0", "doc1", "doc2"]).to_string())

# --- the IDF weights: why the numbers differ from plain counts ---
print()
print("%-10s %-7s %s" % ("word", "idf", "appears in"))
for w, idf in sorted(zip(tfidf_vectorizer.get_feature_names_out(),
                         tfidf_vectorizer.idf_), key=lambda t: t[1]):
    n = sum(1 for d in corpus if w in d.lower())
    print("%-10s %.3f   %d of 3 docs" % (w, idf, n))

# In doc1, "nlp" and "amazing" both have a raw COUNT of 1 - but TF-IDF gives
# amazing 0.652 and nlp 0.385, because nlp is in every document and so tells
# you nothing about which document you are looking at.
#
# Note: sklearn smooths the idf as log((1+n)/(1+df)) + 1, so a word present in
# every document gets 1.0 rather than the textbook 0. Lowest weight, not zero.
#
# Rows are L2-normalised, which is why doc2's five words each show 0.432
# rather than raw weights.
```

```python
print("TF-IDF:\n", X_tfidf.toarray())
print("Vocabulary:", tfidf_vectorizer.get_feature_names_out())
```

### **Exercise 5: N-grams**

Generate `bigrams (2-grams)` from the corpus using `CountVectorizer`. Store the result in `X_bigram`

```python
# your code here
# ngram_range=(2, 2) means: count adjacent PAIRS of words, not single words
bigram_vectorizer = CountVectorizer(ngram_range=(2, 2))
X_bigram = bigram_vectorizer.fit_transform(corpus)

print("bigram vocabulary (%d):" % len(bigram_vectorizer.get_feature_names_out()))
print(" ", list(bigram_vectorizer.get_feature_names_out()))
print()
print(pd.DataFrame(X_bigram.toarray(),
                   columns=bigram_vectorizer.get_feature_names_out(),
                   index=["doc0", "doc1", "doc2"]).to_string())

# --- why bigrams are worth the extra columns: negation ---
# Single words cannot tell these two sentences apart in any useful way,
# because "not" is just one more token sitting beside "good".
negation = ["the film was good", "the film was not good"]

for rng, label in [((1, 1), "unigrams"), ((2, 2), "bigrams ")]:
    v = CountVectorizer(ngram_range=rng)
    v.fit(negation)
    print()
    print("%s vocabulary: %s" % (label, list(v.get_feature_names_out())))

# With bigrams, "not good" becomes its own feature, separate from "was good".
# Local word order is recovered inside a two-word window.
#
# The cost: vocabulary grows fast. Here 9 unigrams -> 8 bigrams on a tiny
# corpus, but on real text the bigram vocabulary is far larger than the
# unigram one, and most bigrams appear only once.
#
# The limit: bigrams catch "not good" but not "not particularly good" - the
# negation is three words away. Trigrams would catch that and miss a four-word
# gap. Full word order needs a different kind of model entirely.
```

```python
print("Bigrams:\n", X_bigram.toarray())
print("Bigram Vocabulary:", bigram_vectorizer.get_feature_names_out())
```

## **4. Advanced Exercise: Custom Preprocessing Pipeline**

### **Exercise 6: Build a Custom Preprocessing Pipeline**

Combine all the preprocessing steps (tokenization, stop word removal, punctuation removal, stemming/lemmatization) into a single function. 

```python
# your code here
def text_preprocessing_pipeline(text):
    # Step 1: Tokenize the text
    # .lower() here so stop-word matching works and the output is consistent
    tokens = word_tokenize(text.lower())

    # Step 2: Remove stop words
    tokens = [w for w in tokens if w not in stop_words]

    # Step 3: Remove punctuation
    tokens = [w for w in tokens if w not in string.punctuation]

    # Step 4: Apply lemmatization
    lemmatized_tokens = [lemmatizer.lemmatize(w) for w in tokens]

    return lemmatized_tokens
```

Apply this function to the following text

```python
text = "Natural Language Processing (NLP) is a fascinating field of study! It involves analyzing and understanding human language."

# your code here
processed_text = text_preprocessing_pipeline(text)

print("raw tokens   :", len(word_tokenize(text)))
print("after pipeline:", len(processed_text))

# a second sentence, where lemmatization earns its keep:
print()
print("The cats are playing with the mice in the garden.")
print(" ->", text_preprocessing_pipeline("The cats are playing with the mice in the garden."))
# cats -> cat, mice -> mouse. A stemmer could never produce "mouse" from
# "mice" - it only chops affixes, and there is no affix to chop.
```

```python
print("Processed Text:", processed_text)
```

## **5. Evaluation of Preprocessing Techniques**

### **Exercise 7: Compare Preprocessing Techniques**

Compare the results of stemming and lemmatization on the following sentence. Store the results in `stemmed_tokens` and `lemmatized_tokens`

```python
sentence = "The cats are playing with the mice in the garden."
# your code here
# Step 1: Tokenize and preprocess the sentence and store the result in filtered_tokens
tokens = word_tokenize(sentence.lower())
tokens = [w for w in tokens if w not in stop_words]
filtered_tokens = [w for w in tokens if w not in string.punctuation]

# Step 2: Apply stemming
stemmed_tokens = [stemmer.stem(w) for w in filtered_tokens]

# Step 3: Apply lemmatization
lemmatized_tokens = [lemmatizer.lemmatize(w) for w in filtered_tokens]

# --- side by side ---
print("%-12s %-12s %s" % ("ORIGINAL", "STEMMED", "LEMMATIZED"))
for o, s, l in zip(filtered_tokens, stemmed_tokens, lemmatized_tokens):
    print("%-12s %-12s %s" % (o, s, l))

# Each technique wins one case here:
#   playing -> stemmer gives "play"  (right), lemmatizer leaves "playing" (wrong)
#   mice    -> stemmer leaves "mice" (wrong), lemmatizer gives "mouse"   (right)
#
# The stemmer fails on "mice" because it only chops affixes and there is no
# affix to chop - an irregular plural needs a dictionary.
# The lemmatizer fails on "playing" only because .lemmatize() assumes the word
# is a noun. Tell it otherwise and it wins both:

print()
print("lemmatizer with the correct part of speech:")
lemmatized_pos = [lemmatizer.lemmatize(w, pos="v") for w in filtered_tokens]
print(" ", lemmatized_pos)
```

```python
print("Original Tokens:", filtered_tokens)
print("Stemmed Tokens:", stemmed_tokens)
print("Lemmatized Tokens:", lemmatized_tokens)
```

### **Exercise 10: Feature Extraction on Tweets**

Extract features from the preprocessed tweets using **Bag of Words** and **TF-IDF**. Store the results in ``X_bow`` and ``X_tfidf``, respectively.

```python
# your code here
# The pipeline returns a LIST OF TOKENS per tweet, but CountVectorizer and
# TfidfVectorizer expect STRINGS - so join the tokens back into a sentence.
joined_tweets = [" ".join(tokens) for tokens in preprocessed_tweets]

# Step 1: Create a Bag of Words representation
bow_vectorizer = CountVectorizer()
X_bow = bow_vectorizer.fit_transform(joined_tweets)

# Step 2: Create a TF-IDF representation
tfidf_vectorizer_tweets = TfidfVectorizer()
X_tfidf = tfidf_vectorizer_tweets.fit_transform(joined_tweets)

print("X_bow  :", X_bow.shape)
print("X_tfidf:", X_tfidf.shape)
print("vocabulary size:", len(bow_vectorizer.get_feature_names_out()))
print("non-zero cells: %.3f%% - the matrix is almost entirely zeros"
      % (100 * X_bow.nnz / (X_bow.shape[0] * X_bow.shape[1])))

# Most of those ~20,000 columns are words appearing in a single tweet, which
# a model cannot learn anything from. min_df drops them:
print()
for m in (1, 2, 5):
    v = CountVectorizer(min_df=m)
    v.fit(joined_tweets)
    print("  min_df=%d -> vocabulary %5d" % (m, len(v.get_feature_names_out())))
```
